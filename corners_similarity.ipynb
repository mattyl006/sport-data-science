```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

```python
import numpy as np
from numpy.linalg import norm

sentence = sentences[0:6]
# vector embeddings created from dataset
embeddings = model.encode(sentence)
#print(embeddings)
# query vector embedding
q = sentences[10]
query_embedding = model.encode(q)
#print(query_embedding)
# define our distance metric
def cosine_similarity(a, b):
    #print("a" ,a)
    #print("b" ,b)
    r = np.dot(a, b)
    result = r/(norm(a)*norm(b))
    return result

# run semantic similarity search
print("Query: ",q)
for e, s in zip(embeddings, sentence):
    # print("e",query_embedding)
    cos = cosine_similarity(e, query_embedding)
    print(s, " -> similarity score = ",
         cosine_similarity(e, query_embedding))
```
