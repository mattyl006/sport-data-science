```python
from get_data.get_data import get_events
import pandas as pd

match = 3836192
def_areas_x = 12
def_areas_y = 8

events_types_to_analize = ['passes', 'carrys', 'ball_receipts', 'ball_recoverys', 'duels', 'dribbles', 'blocks', 'clearances', 
'dispossesseds', 'errors', 'foul_committeds', 'foul_wons', 'half_ends', 'injury_stoppages', 'interceptions', 'miscontrols', 'shots']

events_dfs = []

for event_type in events_types_to_analize:
    events_dfs.append(get_events(event_type, match))
```

```python
def getEventJsonColumnName(event_type):
    return event_type.lower().replace(' ', '_').replace('*', '')
```

```python
types_without_json_column = set()

for event_df in events_dfs:
    event_type = event_df['type'][0]
    print(event_type)
    print(event_df.columns)
    try:
        event_df[getEventJsonColumnName(event_type)][0]
    except:
        types_without_json_column.add(event_type)
        print('does not have json column')
    print()
```

```python
events_dfs_renamed = []

for event_df in events_dfs:
    event_type = event_df['type'][0]
    if event_type not in types_without_json_column:
        events_dfs_renamed.append(event_df.rename(columns={getEventJsonColumnName(event_type) : 'event'}))
    else:
        events_dfs_renamed.append(event_df)
```
