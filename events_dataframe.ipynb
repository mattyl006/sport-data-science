```python
from get_data.get_data import get_events
import pandas as pd

match = 3836192
def_areas_x = 12
def_areas_y = 8

passes = get_events('passes', match)
carrys = get_events('carrys', match)
ball_receipts = get_events('ball_receipts', match)
ball_recoverys = get_events('ball_recoverys', match)
duels = get_events('duels', match)
dribbles = get_events('dribbles', match)
blocks = get_events('blocks', match)
clearances = get_events('clearances', match)
dispossesseds = get_events('dispossesseds', match)
errors = get_events('errors', match)
foul_committeds = get_events('foul_committeds', match)
foul_wons = get_events('foul_wons', match)
half_ends = get_events('half_ends', match)
injury_stoppages = get_events('injury_stoppages', match)
interceptions = get_events('interceptions', match)
miscontrols = get_events('miscontrols', match)
shots = get_events('shots', match)
```

```python
events = pd.concat([passes,
carrys,
ball_receipts,
ball_recoverys,
duels,
dribbles,
blocks,
clearances,
dispossesseds,
errors,
foul_committeds,
foul_wons,
half_ends,
injury_stoppages,
interceptions,
miscontrols,
shots])

events = events.sort_values('index')

events.to_csv('events_merged.csv', encoding='utf-8', index=False)

events
```
