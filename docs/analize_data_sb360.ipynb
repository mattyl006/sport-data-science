# Analiza danych Statbsbomb 360

```python
import statsbombpy
import numpy as np
import pandas as pd
from statsbombpy import sb
```

**Dane 360 z meczu**

```python
match_id = 3800107
match_frames = sb.frames(match_id, fmt='dataframe')
match_frames
```

Dla jednego 'id' istnieje kilka rzędów, każdy z lokalizacją jednego zawodnika z obu drużyn, które są rozróżnione w kolumnie 'teammate' poprzez dwie wartości: *true* lub *false*

# Dane dotyczące jednego eventu

```python
event_id = 'a77d08c5-dfd7-4f50-8cd3-4b6fbf168253'
frames = match_frames.loc[match_frames['id']==event_id]
frames
```

**Lokalizacje zawodników z obu drużyn**

```python
loc_team_1 = list()
loc_team_1.append(frames['location'].loc[frames['teammate']==True])
loc_team_1
```

```python
loc_team_2 = list()
loc_team_2.append(frames['location'].loc[frames['teammate']==False])
loc_team_2
```
