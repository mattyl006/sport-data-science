```python
import pandas as pd
from sequences.get_sequences import get_sequences
```

```python
esa_22_23_matches = pd.read_pickle('esa_22_23_matches_preprocessed.pkl')
df_with_scores = pd.read_pickle('df_with_scores.pkl')
```

```python
esa_22_23_sequences = get_sequences(esa_22_23_matches)
```

```python
dfs_result = []
for id, score_positive, score_negative in zip(df_with_scores['id'].values, df_with_scores['score_positive'].values, df_with_scores['score_negative'].values):
    start_id = id.split('#')[0]
    end_id = id.split('#')[1]
    for seq in esa_22_23_sequences:
        if seq['id'].values[0] == start_id and seq['id'].values[-1] == end_id:
            seq_with_metric = seq
            seq_with_metric['fast_action_score_positive'] = len(seq) * [score_positive]
            seq_with_metric['fast_action_score_negative'] = len(seq) * [score_negative]
            dfs_result.append(seq_with_metric)
fast_actions_with_metric = pd.concat(dfs_result)
fast_actions_with_metric.to_pickle('fast_actions_with_metric.pkl')
fast_actions_with_metric
```
