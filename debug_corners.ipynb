```python
import pandas as pd
from corners.get_dataframe import prepare_matches_from_pkl
from sequences.get_sequences import get_sequences_by_possession, get_corners_sequences
```

```python
corners_evaluated = pd.read_csv('corners_experiment_09062023_181550_evaluated.csv')
```

```python
corners_low_score_and_shot = corners_evaluated[(corners_evaluated['score_positive'] < 0.5) & (corners_evaluated['num_of_shot'] > 0)]
```

```python
corners_high_score_and_no_shot = corners_evaluated[(corners_evaluated['score_positive'] > 0.5) & (corners_evaluated['num_of_shot'] == 0)]
```

```python
corners_low_score_and_shot
```

```python
sequences_ids = corners_low_score_and_shot['id'].values
```

```python
esa_21_22 = prepare_matches_from_pkl('data/esa_21_22_matches.pkl')
esa_22_23 = prepare_matches_from_pkl('data/esa_22_23_matches.pkl')
```

```python
# esa_matches = esa_21_22.append(esa_22_23)
# sequences = get_sequences_by_possession(esa_matches)
# corners = get_corners_sequences(sequences) #FIXME: generates error

sequences_21_22 = get_sequences_by_possession(esa_21_22)
corners_21_22 = get_corners_sequences(sequences_21_22)

sequences_22_23 = get_sequences_by_possession(esa_22_23)
corners_22_23 = get_corners_sequences(sequences_22_23)
```

```python
sequences_to_draw_21_22 = []
sequences_to_draw_22_23 = []

for sequence in corners_21_22:
    id = sequence['id'].values[0] + '-' + sequence['id'].values[-1]
    if id in sequences_ids:
        sequence['score'] = len(sequence) * list(corners_low_score_and_shot[corners_low_score_and_shot['id'] == id]['score_positive'].values)
        sequences_to_draw_21_22.append(sequence)

for sequence in corners_22_23:
    id = sequence['id'].values[0] + '-' + sequence['id'].values[-1]
    if id in sequences_ids:
        sequence['score'] = len(sequence) * list(corners_low_score_and_shot[corners_low_score_and_shot['id'] == id]['score_positive'].values)
        sequences_to_draw_22_23.append(sequence)
```

```python
def prepare_sequences_to_draw(filtered_sequences):
    sequences_to_draw = []
    for sequence in corners_21_22:
        id = sequence['id'].values[0] + '-' + sequence['id'].values[-1]
        if id in sequences_ids:
            sequence['score'] = len(sequence) * list(corners_low_score_and_shot[corners_low_score_and_shot['id'] == id]['score_positive'].values)
            sequences_to_draw.append(sequence)
    for sequence in corners_21_22:
        id = sequence['id'].values[0] + '-' + sequence['id'].values[-1]
        if id in sequences_ids:
            sequence['score'] = len(sequence) * list(corners_low_score_and_shot[corners_low_score_and_shot['id'] == id]['score_positive'].values)
            sequences_to_draw.append(sequence)
    return sequences_to_draw
```

```python
list_corners_low_score_and_shot = prepare_sequences_to_draw(corners_low_score_and_shot)
list_corners_high_score_and_no_shot = prepare_sequences_to_draw(corners_high_score_and_no_shot)
```
