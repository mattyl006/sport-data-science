```python
import pandas as pd
```

```python
fast_actions_evaluated = pd.read_csv('fast_sequences_7.05.14.03_evaluated.csv')
fast_actions_evaluated
```

```python
fast_actions_low_score_and_shot = fast_actions_evaluated[(fast_actions_evaluated['score_negative'] > 0.5) & (fast_actions_evaluated['num_of_shot'] > 0)]
```

```python
sequences_ids = fast_actions_low_score_and_shot['id'].values
print(sequences_ids)
fast_actions_low_score_and_shot
```

```python
esa_21_22 = pd.read_pickle('esa_21_22_matches.pkl')
esa_22_23 = pd.read_pickle('esa_22_23_matches.pkl')
```

```python
from sequences.get_sequences import get_sequences

esa_21_22_sequences = get_sequences(esa_21_22)
```

```python
esa_22_23_sequences = get_sequences(esa_22_23)
```

```python
len(esa_21_22_sequences)
```

```python
sequences_to_draw_1 = []
for seq in esa_21_22_sequences:
    id = seq['id'].values[0] + '-' + seq['id'].values[-1]
    if id in sequences_ids:
        seq['score'] = len(seq) * list(fast_actions_low_score_and_shot[fast_actions_low_score_and_shot['id'] == id]['score_positive'].values)
        sequences_to_draw_1.append(seq)
```

```python
sequences_to_draw_2 = []
for seq in esa_22_23_sequences:
    if (seq['id'].values[0] + '-' + seq['id'].values[-1]) in sequences_ids:
        sequences_to_draw_2.append(seq)
```
