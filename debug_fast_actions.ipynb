```python
esa_21_22_matches = pd.read_pickle('esa_21_22_matches_preprocessed.pkl')
```

```python
esa_21_22_sequences = get_sequences(esa_21_22_matches)
```

```python
fast_actions_evaluated = pd.read_csv('fast_sequences_experiment_11_06_2023_12-33-55_evaluated.csv')
```

```python
fast_actions_evaluated = fast_actions_evaluated.sort_values(by=['score_positive'], ascending=False)
fast_actions_evaluated = fast_actions_evaluated[:10]
```

```python
sequences_ids = list(set(fast_actions_evaluated['id'].values))
sequences_to_draw_1 = []
for seq in esa_21_22_sequences:
    id = seq['id'].values[0] + '#' + seq['id'].values[-1]
    if id in sequences_ids:
        seq['score'] = len(seq) * list(fast_actions_evaluated[fast_actions_evaluated['id'] == id]['score_positive'].values)
        sequences_to_draw_1.append(seq)
```

```python
from draw.Draw import Draw

draw = Draw(scale=2)
start_team = sequences_to_draw_1[0]['possession_team'].values[0]

for i, s in enumerate(sequences_to_draw_1):
    print(f'sequence {i}')
    draw.draw_sequence(s, start_team)
```
