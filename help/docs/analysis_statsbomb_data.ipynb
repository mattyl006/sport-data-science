## Eventy

Gdy wejdziemy w dany mecz otrzymamy słownik z różnymi eventami.

```python
LECH_EXAMPLE_MATCH_ID = 3836192
events = sb.events(match_id=LECH_EXAMPLE_MATCH_ID, split=True, flatten_attrs=False)
events.keys()
```

### Starting xis

Są to dane o składzie danej drużyny oraz ich pozycjach.

```python
starting_xis = events["starting_xis"]
starting_xis
```

```python
print(starting_xis["tactics"][0])
positions_1 = {'G', 'RCB', 'CB', 'LCB', 'RWB', 'LWB', 'CDM', 'RCM', 'LCM', 'RCF', 'LCF'}
```

```python
print(starting_xis["tactics"][1])
positions_2 = {'G', 'RB', 'RCB', 'LCB', 'LB', 'CDM', 'RCM', 'LCM', 'RW', 'LW', 'CF'}
```

```python
for i in positions_1:
    if not (i in positions_2):
        print('positions_1: ' + i)
print()
for j in positions_2:
    if not (j in positions_1):
        print('positions_2: ' + j)
```

Jak widać po powyższym skrypcie układy pozycji różnią się od siebie. W tym przypadku pięcioma pozycjami.
